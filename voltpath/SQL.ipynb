```python
dbutils.widgets.text("catalog", "vp_ms20010501")
catalog = dbutils.widgets.get("catalog")
```

```python
result_q01 = spark.sql(f"""
SELECT
  session_date,
  completed_sessions,
  energy_kwh,
  total_billed,
  total_paid
FROM {catalog}.gold.agg_daily_station
ORDER BY session_date ASC
""")
display(result_q01)
```

```python
result_q02 = spark.sql(f"""
SELECT
  station_id,
  city,
  completed_sessions,
  energy_kwh,
  ROUND(100 * SUM(completed_minutes) / (MAX(num_chargers) * 1440 * COUNT(DISTINCT session_date)), 2) AS utilisation_pct
FROM {catalog}.gold.agg_daily_station
GROUP BY station_id, city, completed_sessions, energy_kwh
ORDER BY utilisation_pct DESC
""")
display(result_q02)
```
